# Bus departures

Edit the assumptions, simulate the selected schedule, then optimize the schedule grid.

```python
from wedding_sim.bus_departures import (
    BehavioralAssumptions, BusDepartureConstraints,
    optimize_behavioral_schedule, simulate_behavioral,
)
from wedding_sim.bus_departures.visualization import (
    plot_behavioral_summary, plot_recommendation,
)
```

```python
constraints = BusDepartureConstraints(guests_riding=94)
assumptions = BehavioralAssumptions()
recommendation = optimize_behavioral_schedule(constraints, assumptions, trials=1_000)
result = simulate_behavioral(constraints, recommendation.schedule, assumptions, trials=10_000)

plot_behavioral_summary(result);
plot_recommendation(constraints, recommendation, result);
```
